```python
# confusion matrix : 

# For a binary classifier, after predicting on test data you get 4 outcomes:

#                         Predicted: Positive         Predicted: Negative

# Actual: Positive        TP (True Positive)          FN (False Negative)
# Actual: Negative        FP (False Positive)         TN (True Negative)

# TP — You said positive, it was positive. Correct.
# TN — You said negative, it was negative. Correct.
# FP — You said positive, actually negative. Type I error.
# FN — You said negative, actually positive. Type II error.

# The matrix tells you where your model fails, not just whether it fails.

# Ex - [[12  0]
#       [ 0  8]]

# it means : 

# 12 → Model predicted class 0, actual was class 0. Correct.
# 8 → Model predicted class 1, actual was class 1. Correct.
# 0 (top right) → Model predicted class 1, actual was class 0. Wrong. → Zero such mistakes.
# 0 (bottom left) → Model predicted class 0, actual was class 1. Wrong. → Zero such mistakes.
```

### 1. Sigmoid Function and Why Logistic Regression Gives Probabilities

Logistic Regression predicts a linear score:

$$
z = w_1x_1 + w_2x_2 + ... + b
$$

Then converts it to probability using sigmoid:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

- Output is always between 0 and 1
- This value is the probability of class 1 (positive class)

```python
# Plot sigmoid curve for intuition
import numpy as np
import matplotlib.pyplot as plt

z = np.linspace(-10, 10, 200)
sigmoid = 1 / (1 + np.exp(-z))

plt.figure(figsize=(8, 4))
plt.plot(z, sigmoid)
plt.axhline(0.5, color='red', linestyle='--', label='0.5 probability')
plt.title('Sigmoid Function')
plt.xlabel('z (linear score)')
plt.ylabel('Probability')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### 2. Decision Threshold (Very Important)

By default, prediction rule is:
- if probability >= 0.5, predict class 1
- else predict class 0

But threshold can be changed based on business goal:
- Lower threshold: catches more positives, but may increase false positives
- Higher threshold: fewer false positives, but may miss true positives

So 0.5 is not always the best choice.

```python
# Try different thresholds and observe confusion matrix changes
from sklearn.metrics import confusion_matrix

proba_class1 = classifier.predict_proba(X_test)[:, 1]

for threshold in [0.3, 0.5, 0.7]:
    y_pred_threshold = (proba_class1 >= threshold).astype(int)
    cm = confusion_matrix(y_test, y_pred_threshold)
    acc = accuracy_score(y_test, y_pred_threshold)
    print(f'\nThreshold = {threshold}')
    print('Accuracy:', round(acc, 4))
    print('Confusion Matrix:\n', cm)
```

### 3. Coefficients, Log-Odds, and Odds Ratio

Logistic regression models **log-odds**:

$$
\log\left(\frac{p}{1-p}\right)=w_1x_1 + w_2x_2 + ... + b
$$

- A positive coefficient increases probability of class 1
- A negative coefficient decreases probability of class 1
- `exp(coefficient)` gives the **odds ratio** for a 1-unit increase in that feature

```python
# Coefficient interpretation table
coef_values = classifier.coef_[0]  # binary classification gives one coefficient vector
coef_table = pd.DataFrame({
    'feature': X.columns,
    'coefficient': coef_values,
    'odds_ratio_exp_coef': np.exp(coef_values)
}).sort_values('coefficient', ascending=False)

coef_table
```

### 4. Regularization (`L1`, `L2`) and `C`

Regularization helps prevent overfitting.

- `L2` (Ridge-like): shrinks coefficients smoothly
- `L1` (Lasso-like): can push some coefficients to zero (feature selection effect)
- `C` is inverse of regularization strength:
  - smaller `C` => stronger regularization
  - larger `C` => weaker regularization

```python
# Compare regularization settings
from sklearn.linear_model import LogisticRegression

settings = [
    ('l2', 0.1),
    ('l2', 1.0),
    ('l2', 10.0),
    ('l1', 0.1),
    ('l1', 1.0),
    ('l1', 10.0),
]

rows = []
for penalty, C in settings:
    model_temp = LogisticRegression(penalty=penalty, C=C, solver='liblinear', random_state=42)
    model_temp.fit(X_train, y_train)
    pred_temp = model_temp.predict(X_test)
    rows.append({
        'penalty': penalty,
        'C': C,
        'test_accuracy': accuracy_score(y_test, pred_temp)
    })

pd.DataFrame(rows).sort_values('test_accuracy', ascending=False)
```

### 5. ROC Curve and AUC

ROC-AUC evaluates ranking quality of predicted probabilities across all thresholds.

- ROC curve: True Positive Rate vs False Positive Rate
- AUC close to 1: strong separability
- AUC around 0.5: near-random model

This is useful when accuracy alone is not enough.

```python
# ROC curve and AUC score
from sklearn.metrics import roc_curve, roc_auc_score

fpr, tpr, _ = roc_curve(y_test, proba_class1)
auc_score = roc_auc_score(y_test, proba_class1)

plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, label=f'ROC curve (AUC = {auc_score:.3f})')
plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Random baseline')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### 6. Assumptions and Practical Tips

Logistic Regression works best when:
- Relationship between features and log-odds is approximately linear
- Features are scaled (important for regularized logistic regression)
- Severe multicollinearity is avoided
- Outliers are handled

Practical notes:
- For imbalanced data, monitor precision/recall/F1, not only accuracy
- Use `class_weight='balanced'` when class imbalance is high
- Tune threshold based on business cost of FP vs FN
- Use cross-validation for robust model selection
