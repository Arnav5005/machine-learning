# How Much Accuracy Is Good?

**Answer:**  
It depends on the problem we are solving.

---

# Problem with Accuracy Metric

Accuracy gives a number like 80%, but it does NOT tell the type of errors being made.

Example: In a binary classification problem (disease prediction)

Model Accuracy = 90% -> Error = 10%

But errors can be of two types:

1. False Negative (FN)  
   Patient has disease -> Model predicts healthy  

2. False Positive (FP)  
   Patient is healthy -> Model predicts diseased  

Accuracy does NOT tell which type of error is happening.

---

# Confusion Matrix

A confusion matrix helps understand error types:

|                | Predicted Positive | Predicted Negative |
|----------------|-------------------|-------------------|
| Actual Positive | TP                | FN                |
| Actual Negative | FP                | TN                |

Accuracy can also be computed as:

Accuracy = (TP + TN) / Total Predictions

---

# Types of Errors

Type-1 Error (False Positive - FP):  
Model predicts positive, but actually negative

Type-2 Error (False Negative - FN):  
Model predicts negative, but actually positive

---

# When Accuracy Is Misleading

Accuracy is misleading in case of an imbalanced dataset.

Example:
If 95% data is negative and model predicts always negative -> 95% accuracy  
But model is useless.

---

# Precision vs Recall

Precision = TP / (TP + FP)  
Recall    = TP / (TP + FN)

---

# Meaning

Precision:  
Out of all predicted positives, how many were actually positive.

Recall:  
Out of all actual positives, how many were correctly identified.

---

# Scenario 1: Email Spam Detection (FP is Dangerous)

Positive class = Spam

False Positive:
A genuine email is predicted as spam

Why dangerous:
- Miss important emails (interview, bank alerts, OTP, etc.)

So we prioritize Precision:
When model says spam, it should be correct.

---

# Scenario 2: Cancer Detection (FN is Dangerous)

Positive class = Cancer

False Negative:
Patient has cancer but model predicts healthy

Why dangerous:
- Treatment delay
- Disease worsens

So we prioritize Recall:
We want to catch as many real cases as possible.

---

# Precision vs Recall (One Line)

Precision -> Controls False Positives (FP)  
Recall -> Controls False Negatives (FN)

---

# Tradeoff (Important)

There is usually a tradeoff between Precision and Recall:

- If you increase Precision, Recall may drop.
- If you increase Recall, Precision may drop.

This happens because changing the decision threshold changes FP and FN differently.

Note: People sometimes say "precision vs accuracy", but in practice the major tradeoff is Precision vs Recall.

---

# F1 Score

F1 Score is the harmonic mean of Precision and Recall.

F1 = 2*(Precision*Recall)/(Precision+Recall)

F1 = 2PR/(P+R)

Why F1 is useful:
- Gives one single score balancing both Precision and Recall.
- Useful when classes are imbalanced.
- Useful when you cannot clearly decide whether FP or FN is more important.

Example where F1 is a good choice:
- Cat-Dog classification (general image classification benchmark)
- If both wrong cat predictions and wrong dog predictions are similarly important, use F1 to balance performance.

---

# Memory Trick

Spam filter -> Don't block important emails -> Precision  
Cancer detection -> Don't miss patients -> Recall  
When both are important -> Use F1 Score
